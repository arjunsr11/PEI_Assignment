### Task 1 : Raw Data Ingestion

Loads each source dataset as-is into the `raw` schema (landing zone):

| Source file | Raw table |
|---|---|
| Customer.xlsx | `raw.customers` |
| Products.csv | `raw.products` |
| Orders.json | `raw.orders` |

All load functions are defined once in the **01_raw_utils** library notebook (`utils/` subfolder) and imported with `%run`.

```python
%run ./utils/01_raw_utils
```

```python
create_schemas()   # raw / enriched / curated schemas (created only if missing)
```

```python
load_raw_cust_data()   # Customer.xlsx -> raw.customers
spark.table("raw.customers").limit(10).display()
```

```python
load_raw_products_data()   # Products.csv -> raw.products (all columns kept as strings)
spark.table("raw.products").limit(10).display()
```

```python
load_raw_orders_data()   # Orders.json -> raw.orders (multiline JSON records)
spark.table("raw.orders").limit(10).display()
```

```python
print("Raw layer row counts:")
for t in ["raw.customers", "raw.products", "raw.orders"]:
    print(f"  {t}: {spark.table(t).count():,}")
```
